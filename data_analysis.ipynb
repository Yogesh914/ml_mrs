```python
import pickle
import pandas as pd
import numpy as np

with open('./data/pre_deployment_data.pkl', 'rb') as f:
    data_dict_pre = pickle.load(f)


with open('./data/post_deployment_data.pkl', 'rb') as f:
    data_dict_post = pickle.load(f)

df_pre = data_dict_pre['merged_df']
df_post = data_dict_post['merged_df']
```

```python
df_pre = df_pre[df_post.columns]

# Filter out any rows where CAPSF1I2s.0 is not 0 or 1
df_pre = df_pre[df_pre['CAPSF1I2s.0'].isin([0, 1])]
df_post = df_post[df_post['CAPSF1I2s.0'].isin([0, 1])]

# Check for duplicate subject IDs
print("Pre-deployment duplicate subjects:", df_pre['subjectID'].duplicated().sum())
print("Post-deployment duplicate subjects:", df_post['subjectID'].duplicated().sum())

# Remove duplicates if any (keeping first occurrence)
df_pre = df_pre.drop_duplicates(subset='subjectID', keep='first')
df_post = df_post.drop_duplicates(subset='subjectID', keep='first')

# Find subjects where CAPSF1I2s.0 value changed between pre and post
pre_ptsd = df_pre.set_index('subjectID')['CAPSF1I2s.0']
post_ptsd = df_post.set_index('subjectID')['CAPSF1I2s.0']

# Find common subjects between pre and post
common_subjects = set(pre_ptsd.index) & set(post_ptsd.index)
print(f"Common subjects: {len(common_subjects)}")

# Vectorized approach to find subjects where PTSD status changed
# Convert set to list for pandas indexing
common_subjects_list = list(common_subjects)

# Align the series to only include common subjects
pre_aligned = pre_ptsd.loc[common_subjects_list]
post_aligned = post_ptsd.loc[common_subjects_list]

# Find where values are different
changed_mask = pre_aligned != post_aligned
changed_subjects = changed_mask[changed_mask].index.tolist()

print(f"Subjects with changed PTSD status: {len(changed_subjects)}")

# Count subjects who changed from 0 to 1 (developed PTSD)
from_0_to_1 = ((pre_aligned == 0) & (post_aligned == 1)).sum()
print(f"Subjects who changed from 0 to 1 (developed PTSD): {from_0_to_1}")

# Count subjects who changed from 1 to 0 (recovered from PTSD)
from_1_to_0 = ((pre_aligned == 1) & (post_aligned == 0)).sum()
print(f"Subjects who changed from 1 to 0 (recovered from PTSD): {from_1_to_0}")

# Create dataframe with post-deployment data for subjects who changed PTSD status
df_changed = df_post[df_post['subjectID'].isin(changed_subjects)].copy()
df_changed.reset_index(inplace=True, drop=True)


df_pre = data_dict_pre['merged_df']
# df_post = data_dict_post['merged_df']

# Remove subjects who changed PTSD status from both pre and post deployment datasets
print(f"Before removal - Pre-deployment shape: {df_pre.shape}")
print(f"Before removal - Post-deployment shape: {df_post.shape}")

# Remove changed subjects from pre-deployment data
df_pre_stable = df_pre[~df_pre['subjectID'].isin(changed_subjects)].copy()
df_pre_stable.reset_index(inplace=True, drop=True)

# Remove changed subjects from post-deployment data
df_post_stable = df_post[~df_post['subjectID'].isin(changed_subjects)].copy()
df_post_stable.reset_index(inplace=True, drop=True)

print(f"After removal - Pre-deployment shape: {df_pre_stable.shape}")
print(f"After removal - Post-deployment shape: {df_post_stable.shape}")
print(f"Removed {len(changed_subjects)} subjects with changed PTSD status")
```

```python
import pickle
import pandas as pd
import numpy as np

with open('pre_deployment_data.pkl', 'rb') as f:
    data_dict_pre = pickle.load(f)


with open('post_deployment_data.pkl', 'rb') as f:
    data_dict_post = pickle.load(f)
```

```python
df_pre = data_dict_pre['merged_df']
df_post = data_dict_post['merged_df']

df_pre = df_pre[df_post.columns]

# Filter out any rows where CAPSF1I2s.0 is not 0 or 1
df_pre = df_pre[df_pre['CAPSF1I2s.0'].isin([0, 1])]
df_post = df_post[df_post['CAPSF1I2s.0'].isin([0, 1])]

#drop all na
df_pre = df_pre.dropna()
df_post = df_post.dropna()

# Check for duplicate subject IDs
print("Pre-deployment duplicate subjects:", df_pre['subjectID'].duplicated().sum())
print("Post-deployment duplicate subjects:", df_post['subjectID'].duplicated().sum())

# Remove duplicates if any (keeping first occurrence)
df_pre = df_pre.drop_duplicates(subset='subjectID', keep='first')
df_post = df_post.drop_duplicates(subset='subjectID', keep='first')

```

```python
# Check data types for non-numeric columns
print("2. DATA TYPE ANALYSIS")
print("-" * 40)

# Get non-numeric columns for df_pre
non_numeric_pre = df_pre.select_dtypes(exclude=[np.number]).columns.tolist()
print(f"Non-numeric columns in df_pre: {non_numeric_pre}")
if non_numeric_pre:
    print("Data types:")
    for col in non_numeric_pre:
        print(f"  {col}: {df_pre[col].dtype}")

# Get non-numeric columns for df_post
non_numeric_post = df_post.select_dtypes(exclude=[np.number]).columns.tolist()
print(f"\nNon-numeric columns in df_post: {non_numeric_post}")
if non_numeric_post:
    print("Data types:")
    for col in non_numeric_post:
        print(f"  {col}: {df_post[col].dtype}")

# Check if there are differences in non-numeric columns between datasets
common_non_numeric = set(non_numeric_pre).intersection(set(non_numeric_post))
pre_only_non_numeric = set(non_numeric_pre) - set(non_numeric_post)
post_only_non_numeric = set(non_numeric_post) - set(non_numeric_pre)

print(f"\nCommon non-numeric columns: {list(common_non_numeric)}")
print(f"Non-numeric columns only in pre: {list(pre_only_non_numeric)}")
print(f"Non-numeric columns only in post: {list(post_only_non_numeric)}")

# Show sample values for non-numeric columns if they exist
if non_numeric_pre:
    print(f"\nSample values from non-numeric columns in df_pre:")
    for col in non_numeric_pre[:3]:  # Show first 3 non-numeric columns
        print(f"  {col}: {df_pre[col].head(3).tolist()}")

if non_numeric_post:
    print(f"\nSample values from non-numeric columns in df_post:")
    for col in non_numeric_post[:3]:  # Show first 3 non-numeric columns
        print(f"  {col}: {df_post[col].head(3).tolist()}")

# Convert non-numeric columns to numeric types
print("\n3. CONVERTING NON-NUMERIC COLUMNS")
print("-" * 40)

# Convert non-numeric columns in df_pre
for col in non_numeric_pre:
    try:
        # Check if values contain decimal points to decide between int and float
        sample_val = str(df_pre[col].dropna().iloc[0]) if not df_pre[col].dropna().empty else "0"
        if '.' in sample_val:
            df_pre[col] = pd.to_numeric(df_pre[col], errors='coerce').astype(float)
            print(f"Converted df_pre['{col}'] to float")
        else:
            df_pre[col] = pd.to_numeric(df_pre[col], errors='coerce').astype(int)
            print(f"Converted df_pre['{col}'] to int")
    except Exception as e:
        print(f"Error converting df_pre['{col}']: {e}")

# Convert non-numeric columns in df_post
for col in non_numeric_post:
    try:
        # Check if values contain decimal points to decide between int and float
        sample_val = str(df_post[col].dropna().iloc[0]) if not df_post[col].dropna().empty else "0"
        if '.' in sample_val:
            df_post[col] = pd.to_numeric(df_post[col], errors='coerce').astype(float)
            print(f"Converted df_post['{col}'] to float")
        else:
            df_post[col] = pd.to_numeric(df_post[col], errors='coerce').astype(int)
            print(f"Converted df_post['{col}'] to int")
    except Exception as e:
        print(f"Error converting df_post['{col}']: {e}")

# Verify conversion
print(f"\nAfter conversion:")
print(f"df_pre non-numeric columns: {df_pre.select_dtypes(exclude=[np.number]).columns.tolist()}")
print(f"df_post non-numeric columns: {df_post.select_dtypes(exclude=[np.number]).columns.tolist()}")

```
